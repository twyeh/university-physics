```python
import sympy
from sympy import symbols, sqrt, simplify

# Define symbolic variables
Q, a, x, y, k = symbols('Q a x y k')

# Position of the positive charge +Q
P1 = sympy.Point(a, 0)

# Position of the negative charge -Q
P2 = sympy.Point(-a, 0)

# Arbitrary point in the x-y plane
P = sympy.Point(x, y)

# --- Electric field due to +Q (E1) ---
# Vector from P1 to P
r1_vec = P - P1
r1_x, r1_y = r1_vec.args[0], r1_vec.args[1]

# Magnitude squared of r1
mag_r1_sq = r1_x**2 + r1_y**2

# Electric field E1 components
E1_x = k * Q * r1_x / mag_r1_sq**(3/2)
E1_y = k * Q * r1_y / mag_r1_sq**(3/2)

# --- Electric field due to -Q (E2) ---
# Vector from P2 to P
r2_vec = P - P2
r2_x, r2_y = r2_vec.args[0], r2_vec.args[1]

# Magnitude squared of r2
mag_r2_sq = r2_x**2 + r2_y**2

# Electric field E2 components
E2_x = k * (-Q) * r2_x / mag_r2_sq**(3/2)
E2_y = k * (-Q) * r2_y / mag_r2_sq**(3/2)

# --- Total Electric Field (E_total) ---
E_total_x = simplify(E1_x + E2_x)
E_total_y = simplify(E1_y + E2_y)

print("電場的 x 分量 (E_x):")
display(E_total_x)

print("\n電場的 y 分量 (E_y):")
display(E_total_y)
```
